```python
%%writefile hello2.cu
#include <stdio.h>

__global__ void mykernel(void) {
    printf("Hello World from GPU thread %d!\n", threadIdx.x);
}

int main() {
    printf("Hello World\n");
    mykernel<<<1, 5>>>();
    cudaDeviceSynchronize();
    return 0;
}
```

```python
!nvcc -arch=sm_75 hello2.cu -o hello2
```

```python
!./hello2
```
